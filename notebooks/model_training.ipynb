# Model training
Runs the training scripts from the notebook (small CPU settings).

```python
import subprocess, sys
def run(cmd): print(subprocess.run(cmd, cwd='../training', capture_output=True, text=True).stdout[-1500:])
run([sys.executable,'train_lstm.py','--epochs','3','--max-train','5000'])
run([sys.executable,'train_gru.py','--epochs','3','--max-train','5000'])
```

```python
run([sys.executable,'train_transformer.py','--model','google/flan-t5-small','--epochs','1','--max-train','2000'])
```

```python
import json, matplotlib.pyplot as plt
for n in ['lstm','gru']:
    h=json.load(open(f'../models/{n}/metrics.json'))['history']
    plt.plot([x['val_loss'] for x in h], label=n)
plt.legend(); plt.xlabel('epoch'); plt.ylabel('val loss'); plt.show()
```
