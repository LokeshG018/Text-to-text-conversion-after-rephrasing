# Data analysis
Run `python training/preprocessing.py` first.

```python
import pandas as pd, json, matplotlib.pyplot as plt
train = pd.read_csv('../data/train/train.csv')
print(json.load(open('../data/processed/stats.json')))
train.head()
```

```python
train['src_len']=train.source.str.split().str.len(); train['tgt_len']=train.target.str.split().str.len()
train[['src_len','tgt_len']].plot.hist(bins=30, alpha=.6); plt.title('Sentence length (words)'); plt.show()
```

```python
import difflib
train['overlap']=[difflib.SequenceMatcher(None,a.lower().split(),b.lower().split()).ratio() for a,b in zip(train.source[:5000],train.target[:5000])]+[None]*(len(train)-5000)
train.overlap.dropna().plot.hist(bins=30); plt.title('Word-sequence overlap source vs target (higher = less diverse)'); plt.show()
```
